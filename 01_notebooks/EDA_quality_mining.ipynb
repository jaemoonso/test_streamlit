이 노트북은 데이터 로딩, 전처리, 기본 시각화까지 빠르게 점검하기 위한 템플릿입니다.

```python
from pathlib import Path
import sys

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parents[0]
SRC = ROOT / "02_src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from data.mine_loader import load_raw_data, parse_mining_dataframe, build_training_table

pd.set_option("display.max_columns", 100)
DATA_PATH = ROOT / "00_data" / "quality_prediction_mining_process" / "MiningProcess_Flotation_Plant_Database.csv"
DATA_PATH
```

```python
raw = load_raw_data(DATA_PATH)
raw.head()
```

```python
df = parse_mining_dataframe(raw)
df = build_training_table(df, target_col="% Silica Concentrate")

df.info()
```

```python
target = "% Silica Concentrate"

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(df[target], kde=True, ax=axes[0])
axes[0].set_title("Target Distribution")

sample = df.sample(min(20000, len(df)), random_state=42)
num_cols = [c for c in sample.columns if c not in ["date"]]
corr = sample[num_cols].corr(numeric_only=True)[target].sort_values(ascending=False)
corr.head(15).plot(kind="bar", ax=axes[1])
axes[1].set_title("Top Correlation with Target")
plt.tight_layout()
```
